```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
```

```python
# indices: number of training cycles N, readout amplitude ampls (first row is amplitudes)
# ampl = 3%
N = np.array([0,2,5,10,20]) # (~)
ampls = np.linspace(0,4.5,num=13) # (%)
# shape: (5, 13)

determ = np.loadtxt("determ_readout_3per.txt", delimiter=',')
print(determ.shape)
random = np.loadtxt("random_readout_3per.txt", delimiter=',')
print(random.shape)
```

```python
# make Fig. 4 a and b

cmap = matplotlib.colormaps['Blues']

fig, ax = plt.subplots(1, 1, figsize=(3.5,2.4))
matplotlib.rc('xtick', labelsize=10)
matplotlib.rc('ytick', labelsize=10)

for ii in range(len(N)):
    plt.plot(ampls, determ[ii,:], 'x-', color=cmap(float(ii+2)/(len(N)+2)), label=str(N[ii]))
plt.xlabel("readout amplitude (%)", fontsize=10)
plt.ylabel(r"$\tilde{C}$", fontsize=10)
plt.axvline(x=3.0, color='k', linestyle='dashed', linewidth=1)
plt.legend(fontsize=10)
ax.set_yticks([0,0.5,1])
plt.tight_layout()
plt.savefig("determ_reaout.pdf")
plt.show()
plt.clf()

fig, ax = plt.subplots(1, 1, figsize=(3.5,2.4))
matplotlib.rc('xtick', labelsize=10)
matplotlib.rc('ytick', labelsize=10)

for ii in range(len(N)):
    plt.plot(ampls, random[ii,:], 'o-', color=cmap(float(ii+2)/(len(N)+2)), label=str(N[ii]))
plt.xlabel("readout amplitude (%)", fontsize=10)
plt.ylabel(r"$\tilde{C}$", fontsize=10)

plt.axvline(x=3.0, color='k', linestyle='dashed', linewidth=1)
plt.legend(fontsize=10)
ax.set_yticks([0,0.5,1])
plt.tight_layout()
plt.savefig("random_readout.pdf")
plt.show()
plt.clf()
```

```python
# indices: number of training cycles N, readout amplitude ampls (first row is amplitudes)
# ampl = 3%
g_t = np.array([2,3,4]) # (~)
print(ampls)
# shape: (5, 13)

determ_all = np.loadtxt("determ_readout_all.txt", delimiter=',')
print(determ.shape)
random_all = np.loadtxt("random_readout_all.txt", delimiter=',')
print(random.shape)
```
